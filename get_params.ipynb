```python
from itertools import product
```

```python
# get params experiment 1

h_trans = [0, 10, 20]
v_trans = [0, 10, 20]
pop_sizes = [25]
modes = ["Sample", "MAP"]
networks = ["fully-connected"]
replace = [True, False]

n_runs = 50

with open("params_ex1.txt", "w") as f:
    for ht, vt, ps, ne, mo, ru, re in product(h_trans, v_trans, pop_sizes, networks, modes, range(n_runs), replace):
        if (ht != 0 or vt != 0):
            f.write(f"{ht} {vt} {ps} {ne} {mo} {ru} {re}\n")
        
```
